```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

```python
file_name = "diabetes.csv"
df = pd.read_csv(file_name)
print("Dataset loaded successfully!")
df.head()
```

```python
print("Dataset shape (rows, columns):", df.shape)
```

```python
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("\nData types and non-null values:")
df.info()
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
print("\nStatistical summary:")
display(df.describe())
```

```python
print("Outcome counts:")
print(df["Outcome"].value_counts())
```

```python
zero_as_missing = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

for column in zero_as_missing:
    if column in df.columns:
        df[column] = df[column].replace(0, np.nan)

print("Missing values after checking invalid zeros:")
print(df.isnull().sum())
```

```python
# Remove rows where the target is missing
df = df.dropna(subset=["Outcome"]).copy()

X = df.drop(columns=["Outcome"])
y = df["Outcome"]

print("Features used:")
print(X.columns.tolist())

print("\nTarget variable: Outcome")
print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training targets:", y_train.shape)
print("Testing targets:", y_test.shape)
```

```python

numerical_columns = X_train.select_dtypes(include=np.number).columns.tolist()
print("Numerical columns:", numerical_columns)

model = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),


    ("scaler", StandardScaler()),

    ("classifier", LogisticRegression(max_iter=1000, random_state=42))
])

print("Pipeline created successfully.")
```

```python
model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

# Calculate accuracy
train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_test_pred)

print(f"Training accuracy: {train_accuracy:.2%}")
print(f"Testing accuracy:  {test_accuracy:.2%}")
```

```python
print("Classification Report:\n")

print(classification_report(
    y_test,
    y_test_pred,
    labels=[0, 1],
    target_names=["No Diabetes", "Diabetes"],
    zero_division=0
))
```

```python
cm = confusion_matrix(y_test, y_test_pred, labels=[0, 1])

print("Confusion matrix:")
print(cm)

```

Conclusion: In the dataset target variable was Outcome, and the remaining columns were used as features.

Missing values were handled using the median for numerical columns. Numerical features were standardized. The data was split into 80% training and 20% testing.

A Logistic Regression model achieved a training accuracy of 79.64% and a test accuracy of 70.78%. For the diabetes class, its precision was 0.60, recall was 0.50, and F1-score was 0.55.
